```python
import torch
from torchvision import models
import torch.nn as nn

# Model A: Fresh/Untrained
model_a = models.resnet18(weights=None)
model_a.fc = nn.Identity()

# Model B: Expert/Pretrained
model_b = models.resnet18(weights=models.ResNet18_Weights.DEFAULT)
model_b.fc = nn.Identity()

# Sanity Check: Are they different?
dummy_input = torch.randn(1, 3, 224, 224)
output_a = model_a(dummy_input)
output_b = model_b(dummy_input)

# If this is close to 1.0, they are too similar to explain differences!
similarity = torch.cosine_similarity(output_a, output_b)
print(f"Representational Similarity: {similarity.item():.4f}")
```
